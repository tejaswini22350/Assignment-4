```python
# 1. Import required libraries
import keras
import pandas
from keras.models import Sequential  # For creating neural network model
from tensorflow.keras.layers import Dense  # For adding dense layers
from sklearn.model_selection import train_test_split  # For splitting dataset
import pandas as pd
import numpy as np

# 2. Load and prepare the dataset
dataset = pd.read_csv('diabetes.csv', header=None).values
# Reads the diabetes dataset and converts it to numpy array
# The dataset has 8 input features and 1 target variable (diabetes: yes/no)

# 3. Split the dataset into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(
    dataset[:,0:8],  # Input features (first 8 columns)
    dataset[:,8],    # Target variable (last column)
    test_size=0.25,  # 25% of data for testing
    random_state=87  # For reproducibility
)

# 4. Set random seed for reproducibility
np.random.seed(123)

# 5. Create the neural network model
my_first_nn = Sequential()

# 6. Add layers to the model
# First dense layer
my_first_nn.add(Dense(20,        # 20 neurons
                      input_dim=8,  # 8 input features
                      activation='relu'))  # ReLU activation function

# Second dense layer
my_first_nn.add(Dense(4,         # 4 neurons
                      activation='relu'))  # ReLU activation function

# Output layer
my_first_nn.add(Dense(1,         # 1 neuron (binary output)
                      activation='sigmoid'))  # Sigmoid for binary classification

# 7. Compile the model
my_first_nn.compile(
    loss='binary_crossentropy',  # Loss function for binary classification
    optimizer='adam',            # Adam optimizer
    metrics=['acc']             # Track accuracy during training
)

# 8. Train the model
my_first_nn_fitted = my_first_nn.fit(
    X_train, Y_train,    # Training data
    epochs=25,           # Number of training iterations
    initial_epoch=0      # Starting epoch
)

# 9. Display model summary
print(my_first_nn.summary())

# 10. Evaluate model on test data
print(my_first_nn.evaluate(X_test, Y_test))
```

```python
# 1. Import required libraries
import keras
import pandas as pd
import numpy as np
from keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_breast_cancer  # Built-in breast cancer dataset
from sklearn.model_selection import train_test_split

# 2. Load the breast cancer dataset
cancer_data = load_breast_cancer()
# This dataset includes:
# - 30 features (measurements of breast cancer tumors)
# - Binary target (malignant or benign)

# 3. Split the dataset into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(
    cancer_data.data,     # Features matrix
    cancer_data.target,   # Target variable
    test_size=0.25,      # 25% for testing
    random_state=87      # For reproducibility
)

# 4. Set random seed for reproducibility
np.random.seed(155)

# 5. Create the neural network model
my_nn = Sequential()

# 6. Add layers to the model
# First (and only hidden) layer
my_nn.add(Dense(20,         # 20 neurons
                input_dim=30,  # 30 input features
                activation='relu'))  # ReLU activation

# Output layer
my_nn.add(Dense(1,          # 1 neuron (binary output)
                activation='sigmoid'))  # Sigmoid for binary classification

# 7. Compile the model
my_nn.compile(
    loss='binary_crossentropy',  # Loss function for binary classification
    optimizer='adam',            # Adam optimizer
    metrics=['acc']             # Track accuracy
)

# 8. Train the model
my_nn_fitted = my_nn.fit(
    X_train, Y_train,   # Training data
    epochs=100,         # Number of training iterations
    initial_epoch=0     # Starting epoch
)

# 9. Display model summary
print(my_nn.summary())

# 10. Evaluate model on test data
print(my_nn.evaluate(X_test, Y_test))
```

```python
# 1. Import necessary libraries
from sklearn.preprocessing import StandardScaler  # For data normalization
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

# 2. Load the breast cancer dataset
data = load_breast_cancer()
X = data.data    # Features
y = data.target  # Target labels (malignant/benign)

# 3. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,     # 20% for testing
    random_state=42    # For reproducibility
)

# 4. Normalize the data using StandardScaler
sc = StandardScaler()
# Fit and transform training data
X_train_normalized = sc.fit_transform(X_train)
# Transform test data (using parameters from training data)
X_test_normalized = sc.transform(X_test)

# 5. Create neural network model
model_normalized = Sequential()

# 6. Add layers with increasing complexity
# First hidden layer
model_normalized.add(Dense(32,                       # 32 neurons
                          activation='relu',         # ReLU activation
                          input_shape=(X_train.shape[1],)))  # Input shape based on feature count

# Second hidden layer
model_normalized.add(Dense(64,              # 64 neurons
                          activation='relu'))

# Third hidden layer
model_normalized.add(Dense(128,             # 128 neurons
                          activation='relu'))

# Output layer
model_normalized.add(Dense(1,               # Binary output
                          activation='sigmoid'))

# 7. Compile the model
model_normalized.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 8. Train the model
model_normalized.fit(
    X_train_normalized, y_train,
    epochs=10,                           # Number of training iterations
    batch_size=32,                       # Samples per gradient update
    validation_data=(X_test_normalized, y_test)  # For monitoring validation performance
)

# 9. Evaluate model and print accuracy
accuracy_normalized = model_normalized.evaluate(X_test_normalized, y_test)[1]
print("Accuracy with normalization:", accuracy_normalized)
```

```python
# 1. Import required libraries
import keras
from keras.datasets import mnist  # MNIST dataset of handwritten digits
from keras.models import Sequential
from keras.layers import Dense, Dropout  # Dropout for regularization
import matplotlib.pyplot as plt  # For plotting results

# 2. Load and prepare MNIST dataset
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 3. Normalize pixel values to range [0,1]
train_images = train_images.astype('float32') / 255
test_images = test_images.astype('float32') / 255

# 4. Define number of classes (digits 0-9)
num_classes = 10

# 5. Convert labels to one-hot encoded format
train_labels = keras.utils.to_categorical(train_labels, num_classes)
test_labels = keras.utils.to_categorical(test_labels, num_classes)

# 6. Create neural network model
model = Sequential()

# 7. Add layers to the model
# First dense layer
model.add(Dense(512,                    # 512 neurons
                activation='relu',       # ReLU activation
                input_shape=(784,)))    # Input shape (flattened 28x28 images)
model.add(Dropout(0.2))                 # 20% dropout for regularization

# Second dense layer
model.add(Dense(512,                    # 512 neurons
                activation='relu'))
model.add(Dropout(0.2))                 # 20% dropout

# Output layer
model.add(Dense(num_classes,            # 10 neurons (one per digit)
                activation='softmax'))   # Softmax for multiclass classification

# 8. Compile the model
model.compile(
    loss='categorical_crossentropy',    # Loss for multiclass classification
    optimizer='adam',                   # Adam optimizer
    metrics=['accuracy']                # Track accuracy
)

# 9. Train the model
history = model.fit(
    train_images.reshape(-1, 784),      # Flatten 28x28 images to 784-pixel vector
    train_labels,
    validation_data=(test_images.reshape(-1, 784), test_labels),
    epochs=20,                          # Number of training iterations
    batch_size=128                      # Samples per gradient update
)

# 10. Plot training results
plt.figure(figsize=(10, 5))

# Plot accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='lower right')

# Plot loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')

plt.show()
```

```python
# 1. Import required libraries
import keras
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout
import matplotlib.pyplot as plt
import numpy as np

# 2. Load and prepare MNIST dataset
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 3. Normalize pixel values to range [0,1]
train_images = train_images.astype('float32') / 255  # Convert to float and scale
test_images = test_images.astype('float32') / 255

# 4. Set up classification parameters
num_classes = 10  # 10 digits (0-9)

# 5. Convert labels to one-hot encoded format
train_labels = keras.utils.to_categorical(train_labels, num_classes)
test_labels = keras.utils.to_categorical(test_labels, num_classes)

# 6. Create neural network model
model = Sequential()

# 7. Add layers to the model
# First dense layer
model.add(Dense(512,                    # 512 neurons
                activation='relu',       # ReLU activation
                input_shape=(784,)))    # Input shape (flattened 28x28 images)
model.add(Dropout(0.2))                 # 20% dropout for regularization

# Second dense layer
model.add(Dense(512,                    # 512 neurons
                activation='relu'))
model.add(Dropout(0.2))                 # 20% dropout

# Output layer
model.add(Dense(num_classes,            # 10 neurons (one per digit)
                activation='softmax'))   # Softmax for multiclass classification

# 8. Compile the model
model.compile(
    loss='categorical_crossentropy',    # Loss for multiclass classification
    optimizer='adam',                   # Adam optimizer
    metrics=['accuracy']                # Track accuracy
)

# 9. Train the model
model.fit(
    train_images.reshape(-1, 784),      # Flatten images to 784-dim vector
    train_labels,
    validation_data=(test_images.reshape(-1, 784), test_labels),
    epochs=20,                          # Number of training iterations
    batch_size=128                      # Samples per gradient update
)

# 10. Visualize a test image
plt.imshow(test_images[0], cmap='gray')  # Display first test image
plt.show()

# 11. Make prediction on the displayed image
prediction = model.predict(test_images[0].reshape(1, -1))  # Reshape for model input
print('Model prediction:', np.argmax(prediction))  # Get predicted digit
```

```python
import keras
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout
import matplotlib.pyplot as plt
import numpy as np

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

train_images = train_images.astype('float32') / 255
test_images = test_images.astype('float32') / 255

num_classes = 10
train_labels = keras.utils.to_categorical(train_labels, num_classes)
test_labels = keras.utils.to_categorical(test_labels, num_classes)

models = []

model = Sequential()
model.add(Dense(512, activation='tanh', input_shape=(784,)))
model.add(Dropout(0.2))
model.add(Dense(num_classes, activation='softmax'))
models.append(('1 hidden layer with tanh', model))

model = Sequential()
model.add(Dense(512, activation='sigmoid', input_shape=(784,)))
model.add(Dropout(0.2))
model.add(Dense(num_classes, activation='softmax'))
models.append(('1 hidden layer with sigmoid', model))

model = Sequential()
model.add(Dense(512, activation='tanh', input_shape=(784,)))
model.add(Dropout(0.2))
model.add(Dense(512, activation='tanh'))
model.add(Dropout(0.2))
model.add(Dense(num_classes, activation='softmax'))
models.append(('2 hidden layers with tanh', model))

model = Sequential()
model.add(Dense(512, activation='sigmoid', input_shape=(784,)))
model.add(Dropout(0.2))
model.add(Dense(512, activation='sigmoid'))
model.add(Dropout(0.2))
model.add(Dense(num_classes, activation='softmax'))
models.append(('2 hidden layers with sigmoid', model))

for name, model in models:
    model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    history = model.fit(train_images.reshape(-1, 784), train_labels, validation_data=(test_images.reshape(-1, 784), test_labels),
                        epochs=20, batch_size=128, verbose=0)
    plt.plot(history.history['loss'], label='train_loss')
    plt.plot(history.history['val_loss'], label='val_loss')
    plt.plot(history.history['accuracy'], label='train_accuracy')
    plt.plot(history.history['val_accuracy'], label='val_accuracy')
    plt.title(name)
    plt.xlabel('Epoch')
    plt.legend()
    plt.show()
loss, accuracy = model.evaluate(test_images.reshape(-1, 784), test_labels, verbose=0)
print('{} - Test loss: {:.4f}, Test accuracy: {:.4f}'.format(name, loss, accuracy))
```
